```python
bronze_df = spark.read.table('workspace.bronze.erp_px_cat_g1v2')
```

# Init

```python
from pyspark.sql.functions import col, trim, when
from pyspark.sql.types import StringType
```

```python
rename_map = {
    'ID' : 'sales_product_id',
    'CAT' : 'product_category',
    'SUBCAT' : 'product_subcategory',
    'MAINTENANCE' : 'product_maintenance'
}
```

# Data Tranfromatons

##Automatically Trimming Whitespaces

```python
for field in bronze_df.schema.fields:
    if isinstance(field.dataType, StringType):
        bronze_df = bronze_df.withColumn(field.name, trim(col(field.name)))
```

## Renaming cryptic columns using the translation dictionary

```python
for oldname, newname in rename_map.items():
    bronze_df = bronze_df.withColumnRenamed(oldname, newname)
```

```python
display(bronze_df)
```

# Writing the code into the silver layer

```python
bronze_df.write \
.mode('overwrite') \
.format('delta') \
.saveAsTable('silver.erp_product_categories')
```
