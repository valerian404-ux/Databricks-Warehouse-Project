# Reading from Bronze Layer

```python
bronze_df = spark.read.table('workspace.bronze.erp_loc_a101')
```

# Init

```python
from pyspark.sql.functions import col, when, trim
from pyspark.sql.types import StringType
```

```python
rename_map = {
    'CID': 'customer_id',
    'CNTRY' :'country'
}
```

#   Data Transformations

## Automatically Trimming Whitespaces

```python
for field in bronze_df.schema.fields:
    if isinstance(field.dataType, StringType):
        bronze_df = bronze_df.withColumn(field.name, trim(col(field.name)))
```

## Renaming cryptic columns using the translation dictionary

```python
for oldname, newname in rename_map.items():
    bronze_df = bronze_df.withColumnRenamed(oldname, newname)
```

# Writing the Data into the silver layer

```python
bronze_df.write \
.mode('overwrite') \
.format('delta') \
.saveAsTable('silver.erp_cust_location')
```
