# Writing the data from the silver layer

```python
silver_df = spark.table('workspace.silver.crm_sales_details')
```

# Business Transformations and modeling

```python
%sql
SELECT
    sd.sales_order_number,
    pr.product_key,
    cu.customer_key,
    sd.sales_order_date,
    sd.sales_ship_date,
    sd.sales_due_date,
    sd.sales_sales,
    sd.sales_quantity,
    sd.sales_price
FROM workspace.silver.crm_sales_details sd
LEFT JOIN workspace.gold.dim_products pr
    ON sd.sales_prduct_key = pr.product_key
LEFT JOIN workspace.gold.dim_customers cu
    ON sd.sales_customer_id = cu.customer_id;
```

# Writing it into the gold layer

```python
silver_df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
