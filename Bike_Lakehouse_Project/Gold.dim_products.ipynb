# Reading from the silver table/s

```python
silver_df = spark.table('workspace.silver.crm_products')
```

# Business transformations and modeling

```python
%sql
SELECT
    ROW_NUMBER() OVER (ORDER BY pn.product_start_date, pn.product_key) AS product_key, -- Surrogate key
    pn.product_id,
    pn.product_key AS product_number,
    pn.product_name,
    pc.sales_product_id AS category_id,
    pc.product_category AS category,
    pc.product_subcategory AS subcategory,
    pc.product_maintenance AS maintenance_flag,
    pn.product_line,
    pn.product_start_date AS start_date
FROM workspace.silver.crm_products pn
LEFT JOIN workspace.silver.erp_product_categories pc
    ON pn.product_key = pc.sales_product_id
--WHERE pn.end_date IS NULL; -- Filter out all historical data
```

# Writing it into the gold layer

```python
silver_df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_products")
```
